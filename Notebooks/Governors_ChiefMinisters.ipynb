```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Define the URLs and corresponding titles
urls = {
    "https://www.india.gov.in/my-government/whos-who/governors": "Governor",
    "https://www.india.gov.in/my-government/whos-who/lt-governors-administrators": "Lt. Governor/Administrator",
    "https://www.india.gov.in/my-government/whos-who/chief-ministers": "Chief Minister"
}

# Lists to store formatted details
Chief_Minister_Names = []
Governor_Names = []
LtGovernor_Names = []
Administrator_Names = []
Chief_Minister_Details = []
Governor_Details = []
LtGovernor_Details = []
Administrator_Details = []

def scrape_table(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    # Locate the table (assuming the first table on the page is the one we need)
    table = soup.find('table')

    if not table:
        print(f"No table found on {url}")
        return None

    # Read table into DataFrame
    df = pd.read_html(str(table))[0]
    return df

def format_table(df, title):
    # Adjust the column indexes based on actual table structure
    for _, row in df.iterrows():
        if title == "Chief Minister":
            sentence = f"The {title} of {row[0]} is {row[1]}."
            Chief_Minister_Names.append(row[1])
            Chief_Minister_Details.append(sentence)
        elif title == "Governor":
            sentence = f"The {title} of {row[0]} is {row[1]}."
            Governor_Names.append(row[1])
            Governor_Details.append(sentence)
        elif title == "Lt. Governor/Administrator":
            if " (Lieutenant Governor)" in row[1]:
                sentence = f"The Lieutenant Governor of {row[0]} is {row[1]}."
                LtGovernor_Names.append(row[1])
                LtGovernor_Details.append(sentence)
            elif "(Administrator)" in row[1]:
                sentence = f"The Administrator of {row[0]} is {row[1]}."
                Administrator_Names.append(row[1])
                Administrator_Details.append(sentence)

# Loop through URLs to scrape and format data
for url, title in urls.items():
    df = scrape_table(url)
    if df is not None:
        format_table(df, title)

# Create formatted sentences for the lists of names
list_chief_ministers = "The List of Names of Chief Ministers of India are: " + ", ".join(Chief_Minister_Names) + "."
list_governors = "The List of Names of Governors of India are: " + ", ".join(Governor_Names) + "."
list_lt_governors = "The List of Names of Lieutenant Governors of India are: " + ", ".join(LtGovernor_Names) + "."
list_administrators = "The List of Names of Administrators of India are: " + ", ".join(Administrator_Names) + "."

# Append these formatted sentences to the details list
Chief_Minister_Details.append(list_chief_ministers)
Governor_Details.append(list_governors)
LtGovernor_Details.append(list_lt_governors)
Administrator_Details.append(list_administrators)
```

```python
# Print all the formatted details
print("Chief Ministers Details:")
for chunk in Chief_Minister_Details:
    print(chunk)
print("-" * 80)

print("Governors Details:")
for chunk in Governor_Details:
    print(chunk)
print("-" * 80)

print("Lieutenant Governors Details:")
for chunk in LtGovernor_Details:
    print(chunk)
print("-" * 80)

print("Administrator Details:")
for chunk in Administrator_Details:
    print(chunk)
print("-" * 80)
```
