# **Supreme Court**

```python
import camelot
import pandas as pd

def process_pdf(url, court_name):
    print(f"Processing {court_name}:\n")

    # Extract the table from the PDF
    tables = camelot.read_pdf(url, pages='all', strip_text='\n', flavor='stream')

    if len(tables) == 0:
        print("No tables found in the PDF.")
        return

    # Since we know there is only one table, we'll focus on it
    table = tables[0]
    df = table.df
    df = df[1:]  # Skip header if necessary

    # Adjusting the DataFrame to remove non-numeric rows at the beginning
    for row in range(len(df)):
        if df.iloc[row, 0].strip().replace(".", "").isdigit():
            break
    df = df[row:]  # Keep rows starting from the first with a numeric value

    # Display the content as meaningful sentences
    format_table(df)

def format_table(df):
    for index, row in df.iterrows():
        name = row[1] if len(row) > 1 else "N/A"
        appointment_date = row[2] if len(row) > 2 else "N/A"
        retirement_date = row[3] if len(row) > 3 else "N/A"
        remarks_text = row[4] if len(row) > 4 else ""

        print(f"Judge  {name}, appointed on {appointment_date}, will retire on {retirement_date}. "
      f"Remarks [Parent High Court]: {remarks_text}.\n")


# PDF URL and court name
pdf_urls = {
    "Supreme Court Judges": "https://cdnbbsr.s3waas.gov.in/s35d6646aad9bcc0be55b2c82f69750387/uploads/2024/08/20240801417686651.pdf"
}

# Process the PDF
for court_name, url in pdf_urls.items():
    process_pdf(url, court_name)
```
