```python
import requests
from bs4 import BeautifulSoup

# List of URLs to scrape
urls = [
    "https://doj.gov.in/national-judicial-data-grid-2/",
    "https://justiceclock.ecourts.gov.in/justiceClock/",
    "https://doj.gov.in/live-streaming-of-court-cases/",
    "https://vcourts.gov.in/virtualcourt/",
    "https://doj.gov.in/annual-reports/",
    "https://doj.gov.in/budget-information/",
    "https://doj.gov.in/law-commission-reports-having-doj-interface/",
    "https://doj.gov.in/division/ecourts/",
    "https://legalaffairs.gov.in/",
    "http://legislative.gov.in/",
    "https://www.indiacode.nic.in/",
    "https://nja.gov.in/",
    "https://nalsa.gov.in/",
    "https://www.indiacode.nic.in/",
    "https://limbs.gov.in/",
    "https://doj.gov.in/terms-and-conditions/",
    "https://doj.gov.in/division/appointments/",
    "https://doj.gov.in/division/access-to-justice/",
    "https://doj.gov.in/division/ecourts/",
    "https://doj.gov.in/division/justice-i/",
    "https://doj.gov.in/division/justice-ii/",
    "https://doj.gov.in/division/administration/",
    "https://doj.gov.in/history/",
    "https://doj.gov.in/about-department/",
    "https://doj.gov.in/about-department/vision-and-mission/",
    "https://doj.gov.in/about-department/function-of-department/",
    "https://doj.gov.in/organization-chart/",
    "https://doj.gov.in/about-department/administrative-setup/",
    "https://doj.gov.in/whos-who/",
    "https://doj.gov.in/monthly-achievements-month-wise/",
    "https://doj.gov.in/citizens-charter/",
    "https://doj.gov.in/acts-and-rules/",
    "https://doj.gov.in/website-information-manager/",
    "https://doj.gov.in/",
    "https://limbs.gov.in/limbs/",
    "https://lawmin.gov.in/",
    "https://amritmahotsav.nic.in/",
    "https://www.g20.org/en/",
    "https://doj.gov.in/allocation-of-business-rules-1961/",
    "https://www.tele-law.in/"
]

def scrape_url(url):
    try:
        # Fetch the content from the URL
        response = requests.get(url)
        html_content = response.text

        # Parse the HTML content using BeautifulSoup
        soup = BeautifulSoup(html_content, "html.parser")

        # Extract text from all <p> tags
        p_text_data = [p.get_text(strip=True) for p in soup.find_all('p')]

        # Find all <div> elements with only text content from its children
        grouped_div_texts = []

        for div in soup.find_all('div'):
                # Collect text from the direct children <div> elements
            child_div_texts = [child.get_text(strip=True) for child in div.find_all('div', recursive=False)]

            # If there are child divs and no further nested divs
            if child_div_texts:
                grouped_sentence = ' '.join(child_div_texts)
                grouped_div_texts.append(grouped_sentence)

        return p_text_data, grouped_div_texts
    except Exception as e:
        print(f"Failed to scrape {url}: {e}")
        return [], []

# Scrape data from all URLs
all_p_text_data = []
all_div_texts = []

for url in urls:
    p_text_data, div_texts = scrape_url(url)
    all_p_text_data.extend(p_text_data)
    all_div_texts.extend(div_texts)

# Print the total number of paragraphs and divs scraped
print(f"Total paragraphs scraped: {len(all_p_text_data)}")
print(f"Total divs scraped: {len(all_div_texts)}")

# Print the scraped data
print("Scraped paragraph data:", all_p_text_data)
print("Scraped div data:", all_div_texts)

all_text_data = all_p_text_data + all_div_texts
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load a pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Convert the scraped text data to embeddings (vectors)
embeddings = model.encode(all_text_data)

# Convert the embeddings to a NumPy array for saving in FAISS
embeddings_array = np.array(embeddings)

# Print the shape of the embeddings array
print(f"Embeddings shape: {embeddings_array.shape}")
```

```python
import faiss

# Dimension of the vectors
d = embeddings_array.shape[1]

# Create a FAISS index (using IndexFlatL2 for simplicity)
index = faiss.IndexFlatL2(d)

# Add the vectors to the index
index.add(embeddings_array)

# Save the index locally
faiss.write_index(index, "vector_database.index")

print("Vector database saved locally.")
```

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

# Load the FAISS index from local storage
index = faiss.read_index("vector_database.index")

# Load the pre-trained model used to create the embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')

# Function to retrieve relevant data using a given prompt
def get_relevant_data(prompt, top_k=5):
    # Encode the prompt into a vector
    query_vector = model.encode([prompt])

    # Convert the query vector to the correct shape for FAISS
    query_vector = np.array(query_vector).astype('float32')

    # Search the FAISS index for the nearest neighbors
    distances, indices = index.search(query_vector, top_k)

    # Retrieve the most relevant data
    relevant_data = [all_text_data[i] for i in indices[0]]

    return relevant_data, distances[0]

# Example prompt to query the vector database
prompt = "How many Registered Cases are there?"

# Get the top 5 most relevant pieces of data
relevant_data, distances = get_relevant_data(prompt, top_k=5)

# Print the relevant data and their respective distances
for i, (data, distance) in enumerate(zip(relevant_data, distances)):
    print(f"Result {i+1} (Distance: {distance}):\n{data}\n")
```

```python
from transformers import pipeline

# Load the question-answering pipeline with the deepset/roberta-base-squad2 model
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

def generate_response(prompt, top_k=5):
    # Retrieve the top-k most relevant data from the vector database
    relevant_data = get_relevant_data(prompt, top_k)

    # Combine the relevant data into a single context
    context = '\n '.join([str(data) for data in relevant_data])

    # Use the QA model to generate an answer from the context
    answer = qa_pipeline(question=prompt, context=context)

    return answer['answer']

# Example user prompt
user_prompt = "How many Registered Cases are there?"

# Generate the response
response = generate_response(user_prompt, top_k=5)

# Print the response
print(response)
```
